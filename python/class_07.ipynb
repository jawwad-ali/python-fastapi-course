```
# Functions

def laptop(ram, color, brand):
  print("These are my specs:", ram, color, brand)

laptop("16GB" , "black" , "HP")
```

```
# Default Parameter

def food(dishname = "daal"):
  print(dishname)

food()
```

```
def username(username = "guest123"):
  print(username)

username("Jamal001")
```

```
# Kwargs - Keywords arguments
# Always defined as the last paramters
# THis will always return as a dictionary

def cart(name, price, **kwargs):
  print(name , price, kwargs)

# cart("TShirt" , 900 , coupon_code="ahmed001", customisation="alijawwad")
cart("TShirt" , 900)
```

```
# Print vs Return statement

def add():
  print(5 + 5)

result = add()
print(result)
```

```
def add():
  return 10 + 11117

response = add()
print(response)
```

# Object Oriented Programming (OOP)

## Class - class is a blueprint or a template of an object.

#### Every class will have an __init__() function

#### Everytime a class is initiated it constructor is called

```
class UserForm:
  def __init__(self, name, age, password):
    # creating the attributes of the class
    self.name = name
    self.age = age
    self.password = password

  def printName(self):
    print(self.name)
    return self.name

  def check_password(self):
    if len(self.password) < 6:
      print("Password length should be valid")
    else:
      print("Password is valid")


my_info = UserForm("Ali", 25, "ABCDEFGHIK")
print(my_info.name)
print(my_info.age)
my_info.check_password()
```

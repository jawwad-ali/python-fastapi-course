```
# Functions

def greet():
  print("Hello")

greet()
```

```
# Static functions
def hello():
  print("Hello")

hello()
```

```
# Dynamic Functions
# Parameters = function creation
# Arguments = Function calling

def welcome(name):
  print(f"Welcome {name}")

  welcome("jamal")
  welcome("Ali")
  welcome("Ahmed")
```

```
# Postitional Arguments

def fighting(blue, red):
  print(f"{blue} is at blue corner {red} is at red corner")

fighting("jamal" , "Ahmed")
```

```
def login(email , password):
  print(f"My email is {email} and my password is {password}")

login("jamal@test.com" , 1243143123)
# login("23213213 , "jamal@test.com")
```

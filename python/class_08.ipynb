```
# Sets in python

saved_videos = {"video_1" , "video_2"}
print(type(saved_videos))
print(saved_videos)
```

```
# Set do not allow duplicates

email_box = {"email_one", "email_two" , "email_one"}
print(email_box)
```

```
# Sets are unordered

jumble_words = {"Laptop" , "camera" , "SSD" }
print(jumble_words[0])
```

```
# User email already exists

registered_users = {"ali@test.com" , "ahmed@test.com", "jamal@test.com"}

new_email = "ahmed@test.com"

if new_email in registered_users:
  print("User already exists")

else:
  print("User Registered")

```

```
# User registration example using Input field

registered_users = {"ali@test.com" , "ahmed@test.com", "jamal@test.com"}

user_email = input("Enter your email")
if user_email in registered_users:
  print("user already exists")

else:
  print("User Successfully Registerd")
```

```
# Union

fruits = {"apple" , "orange" , "banana"}
vegetables = {"potato" , "tomato" , "onion"}

combine = fruits.union(vegetables)
print(combine)
```

```
# Intersection

jamal_friends_list = {"Ali" , "Ahmed" , "Akram" , "Mohd" , "Faizan"}
ali_friends_list = {"Uzair" , "Wahab" , "Ali", "Faizan"}

common = jamal_friends_list.intersection(ali_friends_list)
print(common)
```

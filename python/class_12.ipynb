## Quick Reference

| Concept | Syntax |
|---------|--------|
| Default parameter | `def f(a, b=10):` |
| `*args` | `def f(*items):` → `items` is a **tuple** |
| `**kwargs` | `def f(**opts):` → `opts` is a **dict** |
| Signature order | `def f(a, b=1, *args, **kwargs):` |
| Loop `else` | runs only if the loop was **not** `break`-ed |
| Safe dict read | `d.get(key, default)` |
| Order-preserving dedup | `list(dict.fromkeys(items))` |
| Sort by custom key | `sorted(items, key=lambda x: x[1], reverse=True)` |
| Class attribute | shared by all objects: `ClassName.count += 1` |
| Instance attribute | per-object: `self.name = name` |
| Readable object | `def __str__(self): return "..."` |
| Object with max value | `max(objects, key=lambda o: o.average())` |
| API call | `requests.get(url, params={}, timeout=10).json()` |
